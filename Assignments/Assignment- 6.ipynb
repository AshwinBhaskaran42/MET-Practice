```python
#Build a simple script  to manage tasks using dictionary.
 
#Ask user to enter task 1 name
#Ask user to enter task 2 name
#Ask user to enter task 3 name
#Ask user to update task 2
#Ask user to delete task 3
#print final tasks


task1 = input("Enter task 1 name: ")
task2 = input("Enter task 2 name: ")
task3 = input("Enter task 3 name: ")

tasks={
    1:task1,
    2:task2,
    3:task3
}

updated_task2= input("Enter updated task 2 name: ")
tasks[2] = updated_task2

delete_request =  input("Do you want to delete task 3: ")
if delete_request.lower() == 'yes':
    del tasks[3]

print(tasks)
```

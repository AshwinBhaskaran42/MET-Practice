```python
#1
class TV:
    def __init__(self,brand):
        self.brand=brand
    def turn_on(self):
        print(f"{self.brand} TV is turned on.")
    def turn_off(self):
        print(f"{self.brand} TV is turned off.")

class Remote():
    def __init__(self,tv):
        self.tv=tv

tv1=TV("Samsung")
rem= Remote(tv1)
rem.tv.turn_on()
rem.tv.turn_off()
```

```python
#2
class Engine:
    def __init__(self,eng_type):
        self.eng_type=eng_type

    def start_engine(self):
        print(f"{self.eng_type} Engine turned on")
class Car:
    def __init__(self,eng):
        self.eng=eng

eng1=Engine('v6')
car1=Car(eng1)
car1.eng.start_engine()
    
```

```python
#3
class Room:
    def __init__(self,name,light):
        self.name=name
        self.light=light
        self.light.set_room_name(self.name)

class Light:
    def __init__(self):
        self.room_name= None # writing pass might raise an error. Better to create the object here itself.
    def set_room_name(self, room_name):
        self.room_name= room_name
    def switch_on(self):
        print(f"{self.room_name} light switched on")

l1=Light()
# l1.switch_on()
r1=Room('Room 1',l1)
r1.light.switch_on()

l2=Light()
r2=Room('Room 2', l2)
r2.light.switch_on()
# r1.light.switch_on() # Room 2 light switched on... when r2 is also assigned with l1.. overridden 
```

```python
#4
class Laptop:
    def __init__(self,battery):
        self.battery=battery

class Battery:
    def __init__(self,charge):
        self.charge=charge
    def check_charge(self):
        print(f"Battery Charge: {self.charge}")
bat1=Battery(60)
lap1= Laptop(bat1)
lap1.battery.check_charge()
```

```python
#5
details=[
    {
        "name":"Alex",
        "age":20
    },
    {
        "name":"John",
        "age":25
    }
]

for i,person in enumerate(details, start=1):
    print(f"{i}. Name: {person['name']:<5} Age: {person['age']}")
```

```python
#9
subjects=[]
marks=[]
for i in range(3):
    sub = input("Enter subject name: ")
    mark = input("Enter subject marks: ")
    subjects.append(sub)
    marks.append(mark)

for i in range(3):
    print(f"Subject: {subjects[i]:<10} Marks:{marks[i]}")
```

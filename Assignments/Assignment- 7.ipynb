```python
#1
def greet_user(name):
    return f"Hello {name}"

name = input("Enter name: ")
greeting=greet_user(name)
print(greeting)
```

```python
#2
def check_temperature(temp):
    if temp > 99:
        return "Fever"
    else:
        return "Normal"
temp = int(input("Enter temperature: "))
result = check_temperature(temp)
print(result)
```

```python
#3
def get_last_fruit(input_list):
    return input_list[-1]

fruits = input("Enter a list of fruits seperated with spaces: ").split(" ")
list(fruits)

last_fruit= get_last_fruit(fruits)
print(last_fruit)
```

```python
#4
def get_price_tag(price):
    if price > 1000:
        return "Expensive"
    else:
        return "Affordable"

price = int(input("Enter price: "))
result = get_price_tag(price)
print(result)
```

```python
#6
def uppercase_if_string(string):
    t = type(string)
    if t == str:
         print(string.upper())
    else:
        return "Invalid Input"
 
uppercase_if_string("hello")
uppercase_if_string(3)
```

```python
#7
def safe_divide(num,den):
    if den == 0:
        print("Cannot Divide")
    else:
        return num/den
 
safe_divide(25,0)
safe_divide(25,5)
```

```python
#8


name= input("Enter name: ")
password= input("Enter Password: ")

credentials = {
    'username': name,
    'pass':password
}

def check_login(credentials):

    if credentials.get("pass"):
        print("Login Successful")
    else:
        print("Login Failed")

check_login(credentials)
```

```python
#1. Write a function that accepts any number of numbers using *args and returns the sum of all numbers.

def nums(*num):
    return sum(num)
total= nums(1,2,3)
print(total)
```

```python
#6
def largest_num(*args):
    if not args:
        return None  
    max_num = args[0]
    i = 1
    while i < len(args):
        if args[i] > max_num:
            max_num = args[i]
        i += 1
    return max_num
 
print(largest_num(1, 2, 3, 4,5))
 
```

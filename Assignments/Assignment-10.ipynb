```python
#2
# Base class
class Vehicle:
    def __init__(self, model, speed, fuel_type):
        self.model = model
        self.speed = speed
        self.fuel_type = fuel_type

    def accelerate(self):
        print(f"{self.model} is accelerating at {self.speed} km/h.")

    def stop(self):
        print(f"{self.model} has stopped.")

    def honk(self):
        print(f"{self.model} goes 'Beep Beep!'")

# Car class
class Car(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'Honk Honk!'")

# Bike class
class Bike(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'Meep Meep!'")

# Truck class
class Truck(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'HOOONK!'")

# Create objects
car = Car("Toyota Camry", 180, "Petrol")
bike = Bike("Yamaha R15", 120, "Petrol")
truck = Truck("Volvo FMX", 100, "Diesel")

# Demonstrate usage
print("=== CAR ===")
car.accelerate()
car.honk()
car.stop()

print("\n=== BIKE ===")
bike.accelerate()
bike.honk()
bike.stop()

print("\n=== TRUCK ===")
truck.accelerate()
truck.honk()
truck.stop()
```

```python
#3
class Student:
    def __init__(self,name,rollno,grade,attend):
        self.name=name
        self.roll_no=rollno
        self.grade=grade
        self.attendance = attend
    def introduce(self):
        print(f"Hi, my name is {self.name} and I'm stydying in grade:{self.grade}")

class Teacher:
    def __init__(self, name, id, subject):
        self.name=name
        self.id=id
        self.subject=subject
    def introduce(self):
        print(f"Hello, my name is {self.name} and I teach {self.subject}")

class Classroom:
    def __init__(self, grade, sec, teacher, stu_list):
        self.grade=grade
        self.section=sec
        self.teacher=teacher
        self.students=stu_list
    def class_info(self):
        print(f"Class: {self.grade}{self.section}\nClass Teacher: {self.teacher.name}")
        print(f"Class students: ")
        for st in self.students:
            print(f"{st.name:<10}Roll.No- {st.roll_no:<5}Attendance: {st.attendance} ")

Alex= Student("Alex", 1, 3, 80)
Brian= Student("Brian", 2, 3, 70)
Ceaser= Student("Ceaser", 3, 3, 90)

Jessy= Teacher("Jessy", 1,"English")

class_3A = Classroom(3, 'A', Jessy, [Alex, Brian, Ceaser])

print(class_3A.class_info(),"\n")

print(Alex.introduce())

print(Jessy.introduce())
        


```

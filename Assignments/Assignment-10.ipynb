```python
#2
# Base class
class Vehicle:
    def __init__(self, model, speed, fuel_type):
        self.model = model
        self.speed = speed
        self.fuel_type = fuel_type

    def accelerate(self):
        print(f"{self.model} is accelerating at {self.speed} km/h.")

    def stop(self):
        print(f"{self.model} has stopped.")

    def honk(self):
        print(f"{self.model} goes 'Beep Beep!'")

# Car class
class Car(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'Honk Honk!'")

# Bike class
class Bike(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'Meep Meep!'")

# Truck class
class Truck(Vehicle):
    def honk(self):
        print(f"{self.model} goes 'HOOONK!'")

# Create objects
car = Car("Toyota Camry", 180, "Petrol")
bike = Bike("Yamaha R15", 120, "Petrol")
truck = Truck("Volvo FMX", 100, "Diesel")

# Demonstrate usage
print("=== CAR ===")
car.accelerate()
car.honk()
car.stop()

print("\n=== BIKE ===")
bike.accelerate()
bike.honk()
bike.stop()

print("\n=== TRUCK ===")
truck.accelerate()
truck.honk()
truck.stop()
```

```python
#3
class Student:
    def __init__(self,name,rollno,grade,attend):
        self.name=name
        self.roll_no=rollno
        self.grade=grade
        self.attendance = attend
    def introduce(self):
        print(f"Hi, my name is {self.name} and I'm stydying in grade:{self.grade}")

class Teacher:
    def __init__(self, name, id, subject):
        self.name=name
        self.id=id
        self.subject=subject
    def introduce(self):
        print(f"Hello, my name is {self.name} and I teach {self.subject}")

class Classroom:
    def __init__(self, grade, sec, teacher, stu_list):
        self.grade=grade
        self.section=sec
        self.teacher=teacher
        self.students=stu_list
    def class_info(self):
        print(f"Class: {self.grade}{self.section}\nClass Teacher: {self.teacher.name}")
        print(f"Class students: ")
        for st in self.students:
            print(f"{st.name:<10}Roll.No- {st.roll_no:<5}Attendance: {st.attendance} ")

Alex= Student("Alex", 1, 3, 80)
Brian= Student("Brian", 2, 3, 70)
Ceaser= Student("Ceaser", 3, 3, 90)

Jessy= Teacher("Jessy", 1,"English")

class_3A = Classroom(3, 'A', Jessy, [Alex, Brian, Ceaser])

print(class_3A.class_info(),"\n")

print(Alex.introduce())

print(Jessy.introduce())
        


```

```python
#4
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock
 
    def view_details(self):
        print(f"Product: {self.name}, Price: {self.price}, Stock: {self.stock}")
 
    def update_stock(self, quantity):
        self.stock += quantity
        print(f"Stock for {self.name} updated to {self.stock}")
 
class Cart:
    def __init__(self):
        self.items = []
 
    def add_product(self, product):
        if product.stock > 0:
            self.items.append(product)
            product.stock -= 1
            print(f"{product.name} added to cart.")
        else:
            print(f"{product.name} is out of stock.")
 
    def view_cart(self):
        print("Cart contains:")
        for item in self.items:
            print(f"- {item.name}")
 
class User:
    def __init__(self, username, email, address):
        self.username = username
        self.email = email
        self.address = address
        self.cart = Cart()
 
    def add_to_cart(self, product):
        self.cart.add_product(product)
 
    def view_profile(self):
        print(f"User: {self.username}, Email: {self.email}, Address: {self.address}")
 
product1 = Product("Laptop", 50000, 5)
product2 = Product("Headphones", 2000, 2)
 
user1 = User("john_doe", "john@example.com", "123 Main St")
 
product1.view_details()
user1.view_profile()
user1.add_to_cart(product1)
user1.add_to_cart(product2)
user1.cart.view_cart()
product1.update_stock(10)
```

```python
#5
class Book:
    def __init__(self, title, author, ID):
        self.title = title
        self.author = author
        self.id = ID
        self.issued = False
 
    def read(self):
        print(f"Reading '{self.title}' by {self.author}.")
 
    def issue(self):
        if not self.issued:
            self.issued = True
            print(f"'{self.title}' has been issued.")
        else:
            print(f"'{self.title}' is already issued.")
 
    def return_book(self):
        if self.issued:
            self.issued = False
            print(f"'{self.title}' has been returned.")
        else:
            print(f"'{self.title}' was not issued.")
 
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.books_issued = []
 
    def issue_book(self, book):
        if not book.issued:
            book.issue()
            self.books_issued.append(book)
            print(f"{self.name} has issued '{book.title}'.")
        else:
            print(f"{book.title} is not available.")
 
    def return_book(self, book):
        if book in self.books_issued:
            book.return_book()
            self.books_issued.remove(book)
            print(f"{self.name} has returned '{book.title}'.")
        else:
            print(f"{self.name} does not have '{book.title}'.")
 
class Librarian:
    def __init__(self, name):
        self.name = name
 
    def add_book(self, book, library):
        library.append(book)
        print(f"Librarian {self.name} added '{book.title}' to the library.")
 
    def remove_book(self, book, library):
        if book in library:
            library.remove(book)
            print(f"Librarian {self.name} removed '{book.title}' from the library.")
        else:
            print(f"'{book.title}' not found in the library.")
 
library = []
book1 = Book("1984", "George Orwell", "12345")
book2 = Book("To Kill a Mockingbird", "Harper Lee", "67890")
librarian = Librarian("Ms. Green")
member = Member("Alice", "M001")
 
librarian.add_book(book1, library)
librarian.add_book(book2, library)
print(library[:])
member.issue_book(book1)
book1.read()
member.return_book(book1)
librarian.remove_book(book2,library)
```

```python
a=['cat', 'dog', 'monkey', 'cow']
l=len(a)
# output: catdo, dogmo, monkeyco, cowca
for i in range(l):
    curr_ele=a[i]
    next_ele= a[(i+1)%l] # wrap-around 
                         # When i = 3, then (i + 1) % 4 = 4 % 4 = 0, so a[0] = 'cat'. Works for all values of i, including the last one.
                         # Avoids any IndexError.
    print(curr_ele+next_ele[:2])
```

```python
attempt=[]

score=0
for i,q in enumerate(questions):
    print(f"Question {i+1}: {q['question']}")
    for option in q['options']:
        print(option)
    choice = input("Enter your answer: ")
    attempt_entry=(i,choice, q['answer'])
    if choice.upper() == q['answer']:
        score+=2.0
    else:
        score-=0.5
    attempt.append(attempt_entry)
    print("")

print(f"Your Score out of 10: {score}")

print("Summary:")
print("Question No.    Your Answer    Correct Answer")

for i in attempt:
    index_no, my_ans, correct_ans= i 
    print(f"{index_no}    {my_ans}    {correct_ans}")
```

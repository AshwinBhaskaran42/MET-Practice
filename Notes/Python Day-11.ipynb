```python
# Funtions

a = int(input("Enter first number : "))
b = int(input("Enter second number : "))

def add(num1, num2):
    return num1+num2

def sub(num1, num2):
    return num1-num2

def multiply(num1, num2):
    return num1*num2

def divide(num1, num2):
    return num1/num2

print(add(a,b))
print(sub(a,b))
print(multiply(a,b))
print(divide(a,b))
```

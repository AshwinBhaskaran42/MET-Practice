```python
vowels=['a', 'e', 'i', 'o', 'u']

sen = input("Enter a sentence: ")
vowel_count=0
for charac in sen:
    if charac in vowels:
        vowel_count+=1
print(f"Input sentence: {sen}")
print(f"Vowel count = {vowel_count}")
```

```python
# validpassword criteria -
# 1 uppercase, 1 lowercase, 1 special character, 1 number
import string

def validate_password(pas):
    up=False
    lo=False
    sc=False
    num=False
    for i in pas:
        if i.isupper():
            up=True
        elif i.islower():
            lo=True
        elif i.isdigit():
            num=True
        elif i in string.punctuation:
            sc=True
    return up and lo and num and sc

pas = input("Enter a password: ")
result= validate_password(pas)

if result == True:
    print("Strong Password")
else:
    print("Weak Password")
     
```

```python
# Object Composition
'''
It's a design approach where one class contains objects of another, building functionality by combining simple objects.
Here, the Zoo class has Animal objects, showing a "has-a" relationship, unlike inheritance which defines an "is-a" relationship.
'''



class Animal:
    def __init__(self, name):
        self.name = name
 
class Zoo:
    def __init__(self):
        self.animals = []
animal = Animal("cat")
animal_1 = Animal("Dog")
zoo = Zoo()
zoo.animals.append(animal)
zoo.animals.append(animal_1)
 
print(zoo.animals[1].name)
```

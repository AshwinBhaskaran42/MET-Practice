```python
# Encapsulation - data hiding, controlled access

# self.name --> public
# _ (protected)- works for subclasses also/only
# __(private) - dosen't even work for subclasses
```

```python
# create class Dog
# name, breed
# implement encapsulation for 'name' attribute

class Dog:
    def __init__(self, name, breed):
        self.__name=name 
        self.breed=breed
    
    def set_dog_name(self, name):
        self.__name= name
    def get_dog_name(self):
        return self.__name
    
dog1= Dog('tommy', 'labrador')
# print(dog1.__name) #'Dog' object has no attribute '__name' # since it is private
print(dog1.get_dog_name())
dog1.set_dog_name('jimmy')
print(dog1.get_dog_name())
```

```python
# Inheritance

#class Dog(Animal):
#super().__init__(name, age)

class DefaultUser:
    def __init__(self, name, age=30):
        self.name=name
        self.age=age
    def loading(self):
        print(f"User data is loading. age= {self.age} ")

class User(DefaultUser):
    def __init__(self, name,age):
        self.name=name
        self.age=age
        super().__init__(name,age)

user1= User('John',20)
user1.loading()
```

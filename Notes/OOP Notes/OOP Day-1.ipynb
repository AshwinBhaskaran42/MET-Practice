```python
class Car:
    def __init__(self, brand, color, year):
        self.brand=brand
        self.color=color
        self.year=year
        self.is_broken_down= False
    
    def turn_on(self):
        if not self.is_broken_down:
            print(f"{self.brand} has started")
    
    def turn_off(self):
        if not self.is_broken_down:
            print(f"{self.brand} has turned off")
    
    def move_forward(self, kms):
        if not self.is_broken_down:
            print(f"{self.brand} is moving at {kms}kms/hr")

    def breakdown(self):
        self.is_broken_down= True
        print("Silence..")

car = Car("i10" ,"white", "2010")
print(car.turn_on())
print(car.move_forward(20))
print(car.breakdown())
```
